```python
def calcular_estadisticas_muestra(lista_valores_crudos):

    lista_valores_validos = []

    for elemento in lista_valores_crudos:
        if elemento is not None and elemento >= 0:
            lista_valores_validos.append(float(elemento))

    total_elementos = len(lista_valores_validos)


    if total_elementos == 0:
        return None

    suma_total = sum(lista_valores_validos)
    media = suma_total / total_elementos


    suma_diferencias = 0.0

    for valor in lista_valores_validos:
        diferencia = valor - media
        suma_diferencias += diferencia ** 2

    varianza = suma_diferencias / total_elementos


    desviacion = varianza ** 0.5


    return {
        "media": media,
        "varianza": varianza,
        "desviacion_estandar": desviacion
    }


datos_XYZ = [12.5, -999, 15.0, None, 18.2, 20.1, -5, 14.8]

resultado = calcular_estadisticas_muestra(datos_XYZ)

print("Resultado:", resultado)
```
